# **Problem Statement 1: Basic Flask Welcome Website**

Imagine you want to build a very basic website that displays a welcome message when someone visits it.

## ✅ Your Goal:

* Create a Flask web application in **GitHub Codespaces**.
* When a user visits the homepage `/`, the website should display:

```text
Welcome to My Flask Website!
```

* Run the Flask app using the Codespaces terminal.
* Make port `5000` Public from the **PORTS** tab.
* Open the generated Codespaces URL in browser.


















````python
!pip install flask
from flask import Flask

app = Flask(__name__)

@app.route('/')
def welcome():
    return 'Welcome to My Flask Website!'

if __name__ == '__main__':
    app.run(host = '0.0.0.0',port = 5000)
````
